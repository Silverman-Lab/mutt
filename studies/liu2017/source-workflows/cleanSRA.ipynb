```python
import xml.etree.ElementTree as ET
import pandas as pd

# Path to your XML file (update this to the correct path on your system)
xml_file = 'biosample_result.xml'

# Parse the XML file
tree = ET.parse(xml_file)
root = tree.getroot()

# Initialize a list to store the data
data = []

# Loop through each BioSample
for biosample in root.findall('.//BioSample'):  # Use './/BioSample' to find all BioSample elements
    # Extract attributes from the BioSample tag
    accession = biosample.get('accession')
    
    # Extract IDs
    sample_name_elem = biosample.find(".//Id[@db_label='Sample name']")
    sra_id_elem = biosample.find(".//Id[@db='SRA']")
    
    sample_name = sample_name_elem.text if sample_name_elem is not None else 'N/A'
    sra_id = sra_id_elem.text if sra_id_elem is not None else 'N/A'
    
    # Extract all attributes dynamically
    attributes = {attr.get('attribute_name'): attr.text for attr in biosample.findall('.//Attribute')}
    
    # Create a dictionary for this sample with fixed fields and all attributes
    sample_data = {
        'BioSample': accession,
        'Sample Name': sample_name,
        'accession': sra_id,
        **attributes  # Unpack all attributes into the dictionary
    }
    
    # Append to the data list
    data.append(sample_data)

# Create a DataFrame
df = pd.DataFrame(data)

# Display the first few rows
print(df.head())

df.to_csv('metadata.csv', index=False)
```
